```python
import sympy as sp
```

```python
x = sp.Symbol('x')
```

```python
y=input("Enter The Function use * and ** for operations\n") #In x**3-x-3 format
```

```python
f=sp.sympify(y)
```

```python
print(f)
```

```python
a=float(input("Enter Starting Value\n"))
```

```python
b=float(input("Enter Ending Value\n"))
```

```python
iterations=int(input("Enter Number of Iterations\n"))
```

```python
f1=f.subs(x,a)
```

```python
f2=f.subs(x,b)
```

```python
if f1*f2 >=0:
  print("Bisection Method Fails")
else:
  for i in range(1,iterations+1):
    midpoint=(a+b)/2
    f3=f.subs(x,midpoint)

    if f3==0:
      a=b=midpoint
      break
    if f1*f3<0:
      b=midpoint
    else:
      a=midpoint
  root=(a+b)/2
```

```python
print(root)
```
